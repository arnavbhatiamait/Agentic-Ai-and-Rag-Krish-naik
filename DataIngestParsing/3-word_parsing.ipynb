## Word Doc Implementation

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

DocX2txt Loader

```python
try:
    docx_loader=Docx2txtLoader("data\word_files\proposal.docx")
    docx_documents=docx_loader.load()
    print(f"Loaded {len(docx_documents)} document(s) using Docx2txtLoader.")
    print("type of documents:", type(docx_documents))
    print(docx_documents)
except Exception as e:
    print(f"Error loading Word document with Docx2txtLoader: {e}")
```

Unscturctured Word Document Loader

```python
try:
    docx_loader=UnstructuredWordDocumentLoader("data\word_files\proposal.docx",mode="elements")
    docx_documents=docx_loader.load()
    print(f"Loaded {len(docx_documents)} document(s) using UnstructuredWordDocument    Loader.")
    print("type of documents:", type(docx_documents))
    print(docx_documents)
    for i,doc in enumerate(docx_documents):
        print(f"\nDocument {i+1}:")
        print("Content:", doc.page_content)
        print("Metadata:", doc.metadata)
except Exception as e:
    print(f"Error loading Word document with UnstructuredWordDocumentLoader: {e}")
```
